**Task 3**

```python
number = int(input("Enter a positive number: "))
prime = []

for n in range(1, number + 1):
    count = 0
    for num in range(1, n + 1):
        if n % num == 0:
            count += 1
    if count == 2:
        prime.append(n)
    
print(prime)
```

**Task 5**

```python
class Book:
    def __init__(self, title, author, is_borrowed=False):
        self.title = title
        self.author = author
        self.is_borrowed = is_borrowed

    def to_dict(self):
        return {
            "Title": self.title,
            "Author": self.author,
            "Is Borrowed": self.is_borrowed
        }
    

class Library:
    def __init__(self):
        self.books = []

    def add_book(self):
        title = input("Title: ")
        author = input("Author: ")
        new_book = Book(title, author)
        self.books.append(new_book)

    def borrow_book(self):
        title = input("Enter the title: ")
        for book in self.books:
            if book.title != title or book.is_borrowed == True:
                print("Kitob mavjud emas yoki band")
            elif book.title == title and book.is_borrowed == False:
                book.is_borrowed = True
            
    def return_book(self):
        title = input("Enter the title: ")
        for book in self.books:
            if book.title != title:
                print("Bu kitob bizdan emas!")
            elif book.title == title and book.is_borrowed == True:
                book.is_borrowed = False
    
    def search_by_author(self):
        author = input("Enter the author: ")
        for book in self.books:
            if book.author == author:
                print(book.title)
            
    
    def menu(self):
        while True:
            print("\n--- Menu ---")
            print("1. Add Book")
            print("2. Borrow Book")
            print("3. Return Book")
            print("4. Search by author")
            choice = input("Enter your choice: ")
            if choice == "1":
                self.add_book()
                break
            elif choice == "2":
                self.borrow_book()
                break
            elif choice == '3':
                self.return_book()
                break
            elif choice == '4':
                self.search_by_author()
                break
            else:
                print("Invalid choice. Please try again.")

lib = Library()
lib.menu()
```
