**Step 2: Oddiy string tozalash**
- Har bir string columnning bo‘sh joylarini olib tashlash (strip).
- Bo‘sh stringlarni NaN yoki Null bilan almashtirish.

```python
str_cols = df.select_dtypes(include='str').columns
df[str_cols] = df[str_cols].apply(lambda x: x.str.strip())
df.head()
```

```python
import numpy as np
df = df.replace("", np.nan)
df
```

**Step 3: Raqam va sanalarni tozalash**
- Age, score, GPA, attendance, money_spent kabi numeric columnlarni tozalash va int yoki float formatiga
keltirish.
- Date columnlarni pandas datetime formatiga o‘tkazish.

```python
df[['age', 'score', 'gpa', 'attendance', 'money_spent']]
```

```python
num_map = {
    'twenty': '20',
    'ninety': '90',
    'four point five': '4.5',
    '18 years': '18'
}

for c in ['age', 'score', 'gpa', 'attendance', 'money_spent']:
    s = df[c].astype("str").str.strip().str.lower().replace(num_map)
    s = (
        s.str.replace('usd', '')
        .str.replace('$', '')
        .str.replace(',', '.')
        .str.replace('%', '')
        .str.replace("-", "")
    )

    df[c] = pd.to_numeric(s, errors='coerce')

for c in ['age', 'score', 'attendance']:
    df[c] = df[c].round().astype("Int64")

for c in ['gpa', 'money_spent']:
    df[c] = df[c].astype("Float64")

df.loc[df['gpa'] > 5, 'gpa'] = np.nan
df.loc[df['attendance'] > 100, 'attendance'] = 100

print(df[['age', 'score', 'gpa', 'attendance', 'money_spent']].dtypes)
df[['age', 'score', 'gpa', 'attendance', 'money_spent']].head()
```

```python
df[["date_of_join", "event_time"]]
```

```python
for c in ["date_of_join", "event_time"]:
    
    s = df[c].astype('str').str.strip()
    is_unix = s.str.match(r"^\d{10}$")
    new_col = pd.Series([pd.NaT] * len(df), index=df.index)
    
    new_col[is_unix] = pd.to_datetime(s[is_unix].astype("Int64"), unit='s', errors='coerce')

    new_col[~is_unix] = pd.to_datetime(df.loc[~is_unix, c], format='mixed', errors='coerce')

    df[c] = new_col

df[["date_of_join", "event_time"]].head()
```

**Step 4: Email va phone validation**
- Emaillarni kichik harflarga o‘tkazish va noto‘g‘ri formatdagi emaillarni aniqlash.
- Telefon raqamlarni tozalash va yagona standart formatga keltirish (masalan: +998...).

```python
df['email'] = df['email'].astype('str').str.strip().str.lower()
email_pattern = r"^[a-z0-9._%+-]+@[a-z0-9.-]+\.[a-z]"
valid_email = df['email'].str.match(email_pattern, na=False)
print(df.loc[~valid_email, 'email'].head(10))
df.loc[~valid_email, 'email'] = "Incorrect email format"
```

```python
import re

def clean_phone(x):
    if pd.isna(x):
        return pd.NA
    s = str(x).strip()
    if s == '':
        return pd.NA
    
    s = re.sub(r"x\s*\d+$", '', s, flags=re.IGNORECASE)
    d = re.sub(r"\D", '', s)

    if len(d) == 11 and d.startswith('1'):
        d = d[1:]
    elif len(d) == 13 and d.startswith('001'):
        d = d[3:]
    elif len(d) == 10:
        d = d
    else:
        return pd.NA
    
    return f"+1-{d[:3]}-{d[3:6]}-{d[6:]}"


df['phone'] = df['phone'].apply(clean_phone)
df[['name', 'phone']].head()
```

**Step 5: JSON parsing**
- JSON columnlarini ochish (profile_json).
- JSON ichidagi ma’lumotlarni alohida columnlarga ajratish: hobbies, skills, family, devices.
- JSON ichidagi array yoki nested objectlarni tekshirish va flatten qilish.

```python
df[['name', 'profile_json']]
```

```python
import ast
import json

def parse_json(x):
    if pd.isna(x):
        return None
    s = str(x).strip()
    if s == '' or s == 'INVALID_JSON_DATA':
        return None
    
    try:
        return json.loads(s)
    except:
        pass

    try:
        return ast.literal_eval(s)
    except:
        return None
    
df['profile_obj'] = df['profile_json'].apply(parse_json)

df['hobbies'] = df['profile_obj'].apply(lambda d: d.get('hobbies') if isinstance(d, dict) else None)
df['skills'] = df['profile_obj'].apply(lambda d: d.get('skills') if isinstance(d, dict) else None)
df['family'] = df['profile_obj'].apply(lambda d: d.get('family') if isinstance(d, dict) else None)
df['devices'] = df['profile_obj'].apply(lambda d: d.get('devices') if isinstance(d, dict) else None)

df['hobbies_flat'] = df['hobbies'].apply(lambda x: ', '.join(map(str, x)) if isinstance(x, list) else None)

skills_flat = pd.json_normalize([x if isinstance(x, dict) else {} for x in df['skills']]).add_prefix('skills_')
family_flat = pd.json_normalize([x if isinstance(x, dict) else {} for x in df['family']]).add_prefix('family_')

df = df.explode("devices", ignore_index=True)
devices_norm = pd.json_normalize(
    [x if isinstance(x, dict) else {} for x in df['devices']]
).add_prefix("devices_")
df = pd.concat([df.drop(columns=['devices']), devices_norm], axis=1)


df = pd.concat([df, skills_flat, family_flat], axis=1)


df.head()
```

**Step 6: Address parsing**
- Address columnni alohida qism va key-larga ajratish: shahar, tuman, pochta kodi.
- Raw address columnni saqlab, yangi columnlar hosil qilish (addr_city, addr_district, addr_postal).

```python
import re

def parse_address(address):
    if pd.isna(address):
        return {'city': pd.NA, 'district': pd.NA, 'postal_code': pd.NA}
    
    s = str(address).strip()
    if s == '' or s.upper().startswith('BROKEN'):
        return {'city': pd.NA, 'district': pd.NA, 'postal_code': pd.NA}
    
    # Postal code
    m_zip = re.search(r"\b(\d{6})\b", s)  # guruhlar qivolamiz
    postal = m_zip.group(1) if m_zip else pd.NA  # group(0) bu to'liq cell, group(1) bu birinchi group
    
    # District
    district = pd.NA
    m_dist = re.search(r"\b([a-zA-Z' -]+)\s+(district)\b", s, flags=re.IGNORECASE)  # \b \b bu so'z chegarasi boshqa so'zga ulangan bo'lsa olmaydi
    m_dist_2 = re.search(r"\bUZ\s+\d{6}\s+Tashkent\s+(.+)$", s, flags=re.IGNORECASE)
    m_dist_3 = re.search(r"^(.+)\s+\d+\s*-\s*kv\b", s)
    if m_dist:
        district = m_dist.group(1).strip()
    elif m_dist_2:
        district = m_dist_2.group(1).strip()
    elif m_dist_3:
        district = m_dist_3.group(1).strip()
    
    #shahar
    m_tash = re.search(r"\b(Tashkent)\b", s, flags=re.IGNORECASE)
    if m_tash:
        city = "Tashkent"
    else: #veguldan keyingi oxirgi qismni olamiz
        parts = [p.strip() for p in s.split(',') if p.strip()]
        if parts:
            cand = parts[-1]
            cand = re.sub(r"^\d+s*", "", cand)
            if not re.fullmatch(r"\d{6}", cand) and not cand.upper().startswith('UZ'):
                city = cand

    return {"city": city, 'district': district, 'postal_code': postal}

parsed = df['address_raw'].apply(parse_address).apply(pd.Series)
df['address_city'] = parsed['city']
df['address_district'] = parsed['district']
df['address_postal_code'] = parsed['postal_code']

df[["address_raw", "address_city", "address_district", "address_postal_code"]].head(15)
```

**Step 7: Duplicate va missing data tekshirish**

```python
safe_cols = [c for c in df.columns if not df[c].apply(lambda x: isinstance(x, (dict, list))).any()]

duplicates = df.duplicated(subset=safe_cols, keep='first')
print(duplicates.sum())

df = df.drop_duplicates(subset=safe_cols, keep='first').reset_index(drop=True)
```

**Step 8: Data normalization**
- Gender columnini standard formatga keltirish: Male / Female / Unknown.
- Course columnlarini yagona nomga keltirish: Data Science / Python / Other.
- Status columnlarini standart formatga keltirish (lower-case yoki uniform).

```python
def norm_gender(x):
    g = str(x).strip().lower()

    if g.startswith('f'):
        return 'Female'
    if g.startswith('m'):
        return 'Male'
    else:
        return "Unknown"
    
df['gender'] = df['gender'].apply(norm_gender)
```

```python
c = df["course"].astype("string").str.strip().str.lower()

ds_aliases = [
    "data science", "data_science", "data-science",
    "data sciens", "data_sciense", "ds", "d.s."
]
python_aliases = ["python", "pyhton", "pythno"]

df['course'] = "Other"
df.loc[c.isin(ds_aliases), 'course'] = "Data science"
df.loc[c.isin(python_aliases), 'course'] = "Python"
```

```python
def range_count(df, col, low=None, high=None):
    if col not in df.columns:
        return None
    s = pd.to_numeric(df[col], errors="coerce")
    mask = pd.Series([False] * len(df))
    if low is not None:
        mask = mask | (s < low)
    if high is not None:
        mask = mask | (s > high)
    return int(mask.sum())

print("GPA out-of-range:", range_count(clean, "gpa", 0, 5))
print("Attendance out-of-range:", range_count(clean, "attendance", 0, 100))
print("Score out-of-range:", range_count(clean, "score", 0, 100))
```
