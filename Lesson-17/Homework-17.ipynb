1. **Inner Join on Chinook Database**

```python
import pandas as pd
import sqlite3
```

```python
with sqlite3.connect('data/chinook.db') as con:
    df_customers = pd.read_sql(
        'select * from customers',
        con=con
        )
    df_invoices = pd.read_sql(
        "select * from invoices",
        con=con
    )
```

```python
df_customers.head()
```

```python
df_invoices.head()
```

```python
result = pd.merge(df_customers, df_invoices, on='CustomerId')
```

```python
result.groupby("FirstName")['InvoiceId'].count()
```

```python
def age_group(x):
    if x < 18:
        return 'Child'
    else:
        return "Adult"
    
df_titanic['Age_Group'] = df_titanic['Age'].apply(age_group)
df_titanic
```

```python
def custom(x):
    if x < 60:
        return 'Short'
    elif x < 120:
        return 'Medium'
    else:
        return 'Long'
    
df_movies['category'] = df_movies['duration'].apply(custom)
df_movies
```

Pipeline on Titanic

```python
df_titanic
```

```python
def fill_age(df):
    return df.assign(Age = df['Age'].fillna(df_titanic['Age'].mean()))

def fare_per_age(df):
    return df.assign(fare_per_age = df['Fare'] / df['Age'])

def survived(df):
    return df[df['Survived'] == 1]

df_titanic.pipe(fill_age).pipe(fare_per_age).pipe(survived)
```

Pipeline on Flights

```python
def filter(df):
    return df[df['DEP_DELAY'] > 30]

def per_hour(df):
    return df.assign(Delay_per_hour = (df['DEP_DELAY'] + df['ARR_DELAY']) / df['AIR_TIME'])

df_userdata.pipe(filter).pipe(per_hour)
```
