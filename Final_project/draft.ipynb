```python
import pandas as pd
import numpy as np
import json
```

```python
df = pd.read_csv('super_dirty_students.csv')
copydf = df.copy()
```

```python
import json
import ast

def safe_parse(x):
    try:
        return ast.literal_eval(x)
    except:
        return {}

df['profile_json'] = df['profile_json'].apply(safe_parse)

# print(df['profile_json'])
df['hobbies'] = df['profile_json'].apply(lambda x: x.get('hobbies'))
df['skills'] = df['profile_json'].apply(lambda x: x.get('skills'))
df['family'] = df['profile_json'].apply(lambda x: x.get('family'))
df['devices'] = df['profile_json'].apply(lambda x: x.get('devices'))
df[['hobbies', 'skills', 'family', 'devices']].head()
```
